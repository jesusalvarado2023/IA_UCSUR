```python
import numpy as np
import matplotlib.pyplot as plt

n_lanzamientos = 10000
resultados = []

# Simular lanzamientos de dos dados
for i in range(n_lanzamientos):
    dado1 = np.random.randint(1, 7)
    dado2 = np.random.randint(1, 7)
    resultado = dado1 + dado2
    resultados.append(resultado)

# Crear histograma
valores, bins, patches = plt.hist(resultados, bins=range(2, 14), edgecolor='black', align='left')

# Escribir el número y la frecuencia sobre cada barra
for i in range(len(valores)):
    suma = int(bins[i] + 0.5)       # valor de la suma (2–12)
    frecuencia = int(valores[i])    # frecuencia observada
    plt.text(suma, frecuencia + 0.3,  # posición del texto
             f"{suma}\n({frecuencia})",  # texto mostrado
             ha='center', va='bottom', fontsize=9)

# Título y etiquetas
plt.title(f"Distribución de la suma de dos dados ({n_lanzamientos} lanzamientos)")
plt.xlabel("Resultado de la suma")
plt.ylabel("Frecuencia")
plt.show()
```
